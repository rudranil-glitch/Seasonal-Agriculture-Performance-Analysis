```python

seasons = df_clean['Season'].unique()

season_groups = [
    df_clean[df_clean['Season'] == season]['Yield_Tonnes_Ha']
    for season in seasons
]

anova_stat, p_value = f_oneway(*season_groups)

print("ANOVA F-statistic:", anova_stat)
print("ANOVA p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("\nResult: Yield differs significantly across seasons.")
else:
    print("\nResult: No statistically significant difference in yield across seasons.")
```

```python

r, p = pearsonr(
    df_clean['Rainfall_mm'],
    df_clean['Yield_Tonnes_Ha']
)

print("Pearson correlation coefficient:", r)
print("p-value:", p)

if p < 0.05:
    print("The relationship is statistically significant.")
else:
    print("The relationship is not statistically significant.")
```

```python


Q1 = df_clean['Yield_Tonnes_Ha'].quantile(0.25)
Q3 = df_clean['Yield_Tonnes_Ha'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean['Yield_Tonnes_Ha'] < lower_bound) |
    (df_clean['Yield_Tonnes_Ha'] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of yield outliers:", len(outliers))

display(outliers.head(20))
```

```python

score_data = df_clean[
    ['Yield_Tonnes_Ha',
     'Profit_INR',
     'Water_Efficiency_t_per_1000m3']
].copy()

scaler = StandardScaler()

score_data_scaled = scaler.fit_transform(score_data)

df_clean['Performance_Score'] = score_data_scaled.mean(axis=1)

season_score = df_clean.groupby(
    'Season'
)['Performance_Score'].mean().sort_values(
    ascending=False
)

display(season_score)
```

```python


ml_df = df_clean.copy()


ml_df = ml_df.drop(columns=['Farm_ID'])


ml_df = pd.get_dummies(
    ml_df,
    columns=['State', 'District', 'Crop', 'Season', 'Irrigation_Method'],
    drop_first=True
)

print("ML dataset shape:", ml_df.shape)
```

```python

X = ml_df.drop(columns=['Yield_Tonnes_Ha'])
y = ml_df['Yield_Tonnes_Ha']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

model.fit(X_train, y_train)

predictions = model.predict(X_test)

print("Model trained successfully!")
```

```python

mae = mean_absolute_error(y_test, predictions)

rmse = np.sqrt(
    mean_squared_error(y_test, predictions)
)

r2 = r2_score(y_test, predictions)

print("Mean Absolute Error:", mae)
print("Root Mean Squared Error:", rmse)
print("R² Score:", r2)
```

```python


feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    'Importance',
    ascending=False
)

display(feature_importance.head(20))
```

```python
top_features = feature_importance.head(15)

plt.figure(figsize=(10,7))

sns.barplot(
    data=top_features,
    x='Importance',
    y='Feature'
)

plt.title('Top Factors Associated with Yield Prediction')
plt.xlabel('Importance')
plt.ylabel('Feature')

plt.show()
```
